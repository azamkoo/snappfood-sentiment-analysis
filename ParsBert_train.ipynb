```python


import matplotlib
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import re
from hazm import Normalizer, Lemmatizer
```

```python
df = pd.read_csv('/content/drive/MyDrive/dataset/dataset.csv', sep=None, engine='python')
```

```python
df.head(20)
```

```python
# Initialize Hazm tools
normalizer = Normalizer()
lemmatizer = Lemmatizer()


def clean_text_for_bert(text):
    if not isinstance(text, str):
        return ""

    # 1. Normalization (Keep this, it's essential for Persian)
    text = normalizer.normalize(text)

    # 2. Fix specific noise
    text = re.sub(r'<[^>]+>', ' ', text)  # HTML
    text = re.sub(r'http\S+', ' ', text)  # URLs
    text = re.sub(r'@\w+', ' ', text)  # Mentions

    # 3. Reduce Character Flooding (Keep this)
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)

    # 4. REMOVED: The punctuation/number removal lines.
    # BERT needs "!", "?", and numbers to understand context better.

    return text.strip()
```

```python
# Check if there are any rows where label and label_id mismatch
inconsistent = df[
    ((df['label'] == 'HAPPY') & (df['label_id'] != 0)) |
    ((df['label'] == 'SAD') & (df['label_id'] != 1))
    ]

if not inconsistent.empty:
    print("Warning: Inconsistent labels found!")
    print(inconsistent)
else:
    print("Labels are consistent.")
```

```python
# --- DATA CLEANING PHASE 2: STRUCTURAL FIXES ---

# 1. Fix the messy labels in the 'label' column
cleanup_map = {'0': 'SAD', '1': 'HAPPY'} # اگر دیتاست عددی است
df['label'] = df['label'].replace(cleanup_map)

# 2. Repair the 'label_id' column
# حالا HAPPY کلاس 1 (مثبت) و SAD کلاس 0 (منفی) است
label_to_id = {'SAD': 0, 'HAPPY': 1}
df['label_id'] = df['label'].map(label_to_id)

# 3. Drop the duplicates
df.drop_duplicates(subset=['comment'], inplace=True)

# 4. Drop the useless index column
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])

# 5. Final Sanity Check
print("--- Final Status ---")
print(df['label'].value_counts())
print(f"Any Null label_ids? {df['label_id'].isnull().any()}")
print(f"Total rows: {len(df)}")
```

```python
import os
os.system("pip install transformers datasets accelerate scikit-learn")
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_recall_fscore_support
import torch
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer,
    DataCollatorWithPadding
)
from datasets import Dataset
from transformers import EarlyStoppingCallback
```

```python
# 2. SETUP & CONSTANTS
MODEL_NAME = "HooshvareLab/bert-base-parsbert-uncased"
MAX_LENGTH = 128
BATCH_SIZE = 16
EPOCHS = 3

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")
```

```python
# 3. LOAD & PREPARE DATA
dataset_folder = "/content/drive/My Drive/dataset/"
file_path = os.path.join(dataset_folder, "cleaned_dataset.csv")

# حالا لود کردن واقعی
df = pd.read_csv(file_path)

```

```python
df.head(10)
```

```python
# Rename 'label_id' to 'labels' (Crucial for BERT)
df = df.rename(columns={'label_id': 'labels'})

# Keep only necessary columns and ensure labels are integers
df = df[['cleaned_comment', 'labels']]
df['labels'] = df['labels'].astype(int)

# Split 80/20
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42, stratify=df['labels'])

# Convert to HF Dataset
train_dataset = Dataset.from_pandas(train_df)
test_dataset = Dataset.from_pandas(test_df)
```

```python
# 4. TOKENIZATION
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

def tokenize_function(examples):
    return tokenizer(
        examples["cleaned_comment"],
        padding="max_length",
        truncation=True,
        max_length=MAX_LENGTH
    )

print("Tokenizing data...")
# We remove the text column ('cleaned_comment') so the model only sees numbers
# This prevents the "excessive nesting" error
cols_to_remove = ["cleaned_comment", "__index_level_0__"] # cleaning up artifacts
existing_cols_train = [c for c in cols_to_remove if c in train_dataset.column_names]
existing_cols_test = [c for c in cols_to_remove if c in test_dataset.column_names]

tokenized_train = train_dataset.map(tokenize_function, batched=True, remove_columns=existing_cols_train)
tokenized_test = test_dataset.map(tokenize_function, batched=True, remove_columns=existing_cols_test)
```

```python
# 5. METRICS
# در بخش محاسبه متریک‌ها (Metrics)
def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)

    # چون حالا HAPPY=1 است، متریک‌ها برای کلاس مثبت (Happy) محاسبه می‌شوند
    precision, recall, f1, _ = precision_recall_fscore_support(labels, predictions, average='binary')
    acc = accuracy_score(labels, predictions)
    return {
        'accuracy': acc,
        'f1': f1,
        'precision': precision,
        'recall': recall
    }
```

```python
# 6. INITIALIZE MODEL
model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=2
).to(device)
```

```python

EPOCHS = 10

# 7. TRAINER SETUP
training_args = TrainingArguments(
    output_dir="./results",
    learning_rate=2e-5,
    per_device_train_batch_size=BATCH_SIZE,
    per_device_eval_batch_size=BATCH_SIZE,
    num_train_epochs=EPOCHS,
    weight_decay=0.01,
    eval_strategy="epoch",        # ارزیابی در پایان هر اپوک
    save_strategy="epoch",        # ذخیره در پایان هر اپوک (الزامی برای EarlyStopping)
    load_best_model_at_end=True,  # لود کردن بهترین مدل در پایان آموزش
    metric_for_best_model="f1",   # معیار سنجش بهترین مدل (می‌توانی accuracy هم بگذاری)
    greater_is_better=True,       # چون f1 هر چه بالاتر باشد بهتر است
    fp16=True,
    report_to="none"
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train,
    eval_dataset=tokenized_test,
    tokenizer=tokenizer,
    data_collator=DataCollatorWithPadding(tokenizer=tokenizer),
    compute_metrics=compute_metrics,
    # اضافه کردن کال‌بک Early Stopping
    callbacks=[EarlyStoppingCallback(early_stopping_patience=2)]
   # patience=2 یعنی اگر تا 2 اپوک متوالی مِتریک بهبود نیافت، آموزش را متوقف کن
)

# 8. TRAIN & SAVE
print("Starting training with Early Stopping...")
trainer.train()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report

# گرفتن پیش‌بینی‌ها از مدل
predictions = trainer.predict(tokenized_test)
preds = np.argmax(predictions.predictions, axis=-1)
labels = predictions.label_ids

# چاپ گزارش کامل (Precision, Recall, F1)
print(classification_report(labels, preds, target_names=['SAD', 'HAPPY']))
```

```python
cm = confusion_matrix(labels, preds)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['SAD', 'HAPPY'],
            yticklabels=['SAD', 'HAPPY'])
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.title('Confusion Matrix - ParsBERT')
plt.show()
```

```python
import torch.nn.functional as F

# تبدیل Logits به احتمالات (Softmax)
probs = F.softmax(torch.tensor(predictions.predictions), dim=-1).numpy()
happy_probs = probs[:, 1] # احتمال خوشحال بودن

plt.figure(figsize=(10, 6))
sns.histplot(happy_probs, bins=20, kde=True, color='green')
plt.axvline(0.5, color='red', linestyle='--')
plt.title('Distribution of Model Confidence (Probability of HAPPY)')
plt.xlabel('Probability')
plt.ylabel('Frequency')
plt.show()
```

```python
sentences = [
    "واقعا خسته نباشید با این سرویس‌دهی ضعیفتون!",
    "بد نبود، ولی می‌تونست خیلی بهتر باشه.",
    "فکر نمی‌کردم انقدر زود برسه، عالی بود.",
    "حیف اون پولی که بابت این غذا دادم."
]

# تنظیم مدل روی حالت ارزیابی با نام متغیر صحیح شما
model.eval()

for s in sentences:
    # تبدیل متن به اعداد (Tokenization)
    inputs = tokenizer(s, return_tensors="pt", truncation=True, padding=True, max_length=128).to(device)

    with torch.no_grad():
        output = model(**inputs)
        # محاسبه احتمالات
        prob = torch.nn.functional.softmax(output.logits, dim=-1)
        happy_score = prob[0][1].item()

        print(f"Sentence: {s}")
        print(f"Score: {happy_score:.4f} ({'HAPPY' if happy_score > 0.5 else 'SAD'})")
        print("-" * 30)
```
